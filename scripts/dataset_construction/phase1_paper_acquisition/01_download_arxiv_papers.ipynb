```python
# =============================================================================
# 功能：批量下载 arXiv 论文的源代码（tar.gz）和 PDF 文件
#       1. 设置搜索参数（日期范围、最大数量），调用 arXiv API 获取论文 ID 列表
#       2. 将 ID 列表保存为文本文件
#       3. 根据 ID 列表逐篇下载：优先下载 tar.gz 源码包，同时下载 PDF；
#          若返回 Content-Type 为 PDF 则仅下载 PDF，并统计各类型数量
#       4. 在指定目录下按论文 ID 创建子文件夹，存放下载的文件
# =============================================================================
from arxiv_download_backend import *
import arxiv
import numpy as np
File_Name='ARXIV_100'
quest_number=150

#按文章分类搜索文章号
client = arxiv.Client(
    page_size=1000,  # 每次请求的最大返回结果数
    delay_seconds=3  # 请求之间的延迟（防止速率限制）
)
# search = arxiv.Search(
#     query="(cat:physics.* ANDNOT cat:physics.optics) OR (cat:cs.* ANDNOT cat:cs.AI ANDNOT cat:cs.CL ANDNOT cat:cs.CV) OR cat:math.* OR cat:stat.* OR cat:q-bio.* OR cat:chem.* OR cat:engineering.* OR cat:econ.* AND submittedDate:[202502010000 TO 202503200000]",  # 指定分类为光学physics.optics   cs.AI 人工智能  cs.CL 计算和语言 cs.CV 计算机视觉和模式识别
#     max_results=200,  # 设置返回的最大论文数量
#     sort_by=arxiv.SortCriterion.SubmittedDate # 按提交日期排序
# )
search = arxiv.Search(
    #query="(cat:physics.* ) AND submittedDate:[202502010000 TO 202503200000]",  # 指定分类为光学physics.optics   cs.AI 人工智能  cs.CL 计算和语言 cs.CV 计算机视觉和模式识别
    query=" submittedDate:[202502010000 TO 202503200000]",  # 指定分类为光学physics.optics   cs.AI 人工智能  cs.CL 计算和语言 cs.CV 计算机视觉和模式识别
    max_results=quest_number,  # 设置返回的最大论文数量
    sort_by=arxiv.SortCriterion.SubmittedDate # 按提交日期排序
)
results = list(client.results(search))
if not results:
    print("没有找到结果。")
else:
    print(f"找到的论文数量: {len(results)}")
Arxiv_lists = []
for result in results:
    entry_id = result.entry_id.split('/')[-1]  # 提取 entry_id 的最后部分
    Arxiv_lists.append(entry_id)  # 将 URL 添加到列表中

#保存文章号
ARXIV_number='data/archive/initial_collection/ARXIV.txt'
os.makedirs(os.path.dirname(ARXIV_number), exist_ok=True)
np.savetxt(ARXIV_number, Arxiv_lists,fmt='%s')

#根据txt下载压缩包
ARXIV_download= np.loadtxt(ARXIV_number, dtype=str).tolist()
ARXIV_download = [i.strip() for i in ARXIV_download if len(i) > 0]

try:
    # 创建下载文件夹
    base = 'data/archive/initial_collection/'+File_Name
    make_dir_if_not_exist(base)

    inp = os.path.join(base, 'input')  # 输入文件夹
    make_dir_if_not_exist(inp)
    out = os.path.join(base, 'output')  # 输出文件夹
    make_dir_if_not_exist(out)
    npdf=0
    ntar=0

    for i, Arxiv_link in enumerate(ARXIV_download):
        source_link = f"http://arxiv.org/e-print/{Arxiv_link}"  # 生成下载链接
        pdf_link = f"https://arxiv.org/pdf/{Arxiv_link}.pdf"
        # 下载tar.gz文件
        try:
            response = requests.get(source_link, timeout=(10, 20))
            content_type = response.headers.get('Content-Type')
            if content_type == 'application/pdf':
                npdf+=1
                print(f"{i+1}:{Arxiv_link} 是第{npdf}个pdf文件")
            elif content_type == 'application/x-tar'or content_type =='application/gzip':
                ntar+=1
                article_folder = os.path.join(inp, Arxiv_link)
                make_dir_if_not_exist(article_folder)
                filename = f"{Arxiv_link}.tar.gz"
                filepath = os.path.join(article_folder, filename)
                # 将下载的内容写入文件
                open(filepath, "wb").write(response.content)

                pdf_response = requests.get(pdf_link, timeout=(10, 20))
                if pdf_response.status_code == 200:
                    pdf_filename = f"{Arxiv_link}.pdf"
                    pdf_filepath = os.path.join(article_folder, pdf_filename)
                    with open(pdf_filepath, "wb") as f:
                        f.write(pdf_response.content)
                print(f"{i+1}:{Arxiv_link} 是第{ntar}个tar+pdf文件")
            else:
                print(f"{i+1}: {Arxiv_link} 出错，未识别的Content-Type: {content_type}")
        except requests.exceptions.Timeout:
                print(f"{i+1}: {Arxiv_link} 下载超时，跳过该文件")
        except Exception as e:
                print(f"{i+1}: {Arxiv_link} 出错：{str(e)}")
    print(f"所有论文下载完成，zip文件路径：{base}\n{len(ARXIV_download)}篇文章中{npdf}个只有pdf，{ntar}个有latex+pdf。")
except Exception as e:
    print(f"下载过程中出现错误：{str(e)}")
```
