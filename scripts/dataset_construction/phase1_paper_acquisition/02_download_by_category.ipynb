```python
# =============================================================================
# 功能：根据arXiv官方分类（如cs.LG、stat.AP等），按照自定义的映射分组（大类/子类），
#       自动下载指定时间范围内每类论文的源代码（tar.gz）和PDF，并为每篇文章生成元数据。
#       主要流程包括：
#       1. 定义官方小类到自定义分类的映射关系；
#       2. 为每个官方小类随机分配目标下载数量（3或4篇）；
#       3. 按顺序遍历每个官方小类，先按日期范围搜索，若不足则回退不限时间；
#       4. 对每篇候选文章去重、下载tar和PDF，并保存到对应的分组目录；
#       5. 生成全局元数据CSV和汇总JSON，记录每篇文章的编号、标题、作者、分类等信息。
# =============================================================================
import os
import re
import csv
import json
import shutil
import tarfile
import requests
import arxiv
from pathlib import Path
from collections import defaultdict
import time
import random

# =========================================================
# 1. 基本配置
# =========================================================
FILE_NAME = "ARXIV_BY_CATEGORY"

# 时间范围：先在这个范围内抓；若某个官方小类不足目标数量，再回退到不限时间
DATE_START = "202506010000"
DATE_END = "202512150000"

# 每个官方小类的目标数量：随机 3 或 4，平均 3.5
OFFICIAL_TARGET_CHOICES = (9, 10)
RANDOM_SEED = 20260316

# 单个官方小类最多拉多少候选结果
MAX_RESULTS_PER_OFFICIAL_CATEGORY = 120

# arXiv 客户端配置
CLIENT_PAGE_SIZE = 100
CLIENT_DELAY_SECONDS = 3

# 单篇文章总处理时限（下载 tar+pdf + 基本校验）
ARTICLE_TIME_LIMIT_SECONDS = 20

# 请求头
HEADERS = {
    "User-Agent": "Mozilla/5.0 (compatible; arxiv-category-downloader/2.0)"
}


# =========================================================
# 2. 工具函数
# =========================================================
def make_dir_if_not_exist(path: Path):
    path.mkdir(parents=True, exist_ok=True)


def sanitize_name(name: str) -> str:
    """
    把文件夹名中不适合 Windows / Linux 的字符替换掉
    """
    name = re.sub(r'[\\/:*?"<>|]', "_", name)
    return name.strip()


def get_arxiv_id_from_result(result) -> str:
    """
    从 arxiv 返回结果中提取文章 id，并去掉版本号
    例如：
    http://arxiv.org/abs/2501.12345v2 -> 2501.12345
    """
    raw_id = result.entry_id.split("/")[-1]
    raw_id = re.sub(r"v\d+$", "", raw_id)
    return raw_id


def normalize_mapping(raw_mapping: dict) -> dict:
    """
    把映射字典标准化为：
    {
        "cs.LG": {
            "custom_subclass": "...",
            "big_class": "Computer Science"
        },
        ...
    }
    """
    normalized = {}
    for official_cat, value in raw_mapping.items():
        custom_subclass, big_class = value
        normalized[official_cat] = {
            "custom_subclass": custom_subclass,
            "big_class": big_class
        }
    return normalized


def build_group_schema(normalized_mapping: dict):
    """
    按 (大类, 自定义子类) 分组，
    得到每个目录类别下包含哪些官方小类
    """
    groups = defaultdict(list)
    for official_cat, info in normalized_mapping.items():
        key = (info["big_class"], info["custom_subclass"])
        groups[key].append(official_cat)

    for key in groups:
        groups[key] = sorted(groups[key])

    return dict(groups)


def build_query(official_cat: str, use_date_range: bool = True) -> str:
    """
    构造 arXiv 查询语句
    """
    if use_date_range:
        return f'cat:{official_cat} AND submittedDate:[{DATE_START} TO {DATE_END}]'
    return f'cat:{official_cat}'


def build_official_targets(normalized_mapping: dict):
    """
    为每个官方小类随机分配 3 或 4 篇目标数
    """
    rng = random.Random(RANDOM_SEED)
    targets = {}
    for official_cat in sorted(normalized_mapping.keys()):
        targets[official_cat] = rng.choice(OFFICIAL_TARGET_CHOICES)
    return targets


def is_tar_like_response(response: requests.Response) -> bool:
    """
    判断 e-print 返回内容是否是 tar.gz 源码包
    """
    content_type = response.headers.get("Content-Type", "").lower()

    if "application/x-tar" in content_type:
        return True
    if "application/gzip" in content_type:
        return True
    if "application/x-gzip" in content_type:
        return True

    if "application/octet-stream" in content_type:
        content = response.content
        if len(content) >= 2 and content[:2] == b"\x1f\x8b":
            return True

    tmp_path = None
    try:
        import tempfile
        with tempfile.NamedTemporaryFile(delete=False, suffix=".tar.gz") as f:
            f.write(response.content)
            tmp_path = f.name
        if tarfile.is_tarfile(tmp_path):
            return True
    except Exception:
        pass
    finally:
        if tmp_path and os.path.exists(tmp_path):
            os.remove(tmp_path)

    return False


def download_tar_and_pdf(session: requests.Session, arxiv_id: str, save_dir: Path):
    """
    下载 tar.gz 和 pdf。
    单篇文章总耗时超过 ARTICLE_TIME_LIMIT_SECONDS 直接跳过。
    只有两者都成功才返回 True。
    """
    eprint_url = f"https://arxiv.org/e-print/{arxiv_id}"
    pdf_url = f"https://arxiv.org/pdf/{arxiv_id}.pdf"

    start_time = time.monotonic()

    def remaining_seconds():
        return ARTICLE_TIME_LIMIT_SECONDS - (time.monotonic() - start_time)

    try:
        remain = remaining_seconds()
        if remain <= 0:
            return False, "article time budget exceeded before e-print"

        eprint_resp = session.get(
            eprint_url,
            timeout=max(1, min(remain, ARTICLE_TIME_LIMIT_SECONDS)),
            allow_redirects=True
        )
        if eprint_resp.status_code != 200:
            return False, f"e-print status_code={eprint_resp.status_code}"

        if not is_tar_like_response(eprint_resp):
            return False, "not tar+gzip source"

        remain = remaining_seconds()
        if remain <= 0:
            return False, "article time budget exceeded before pdf"

        pdf_resp = session.get(
            pdf_url,
            timeout=max(1, min(remain, ARTICLE_TIME_LIMIT_SECONDS)),
            allow_redirects=True
        )
        if pdf_resp.status_code != 200:
            return False, f"pdf status_code={pdf_resp.status_code}"

        pdf_content_type = pdf_resp.headers.get("Content-Type", "").lower()
        if "pdf" not in pdf_content_type:
            return False, f"pdf content-type invalid: {pdf_content_type}"

        remain = remaining_seconds()
        if remain <= 0:
            return False, "article time budget exceeded before write"

        make_dir_if_not_exist(save_dir)

        tar_path = save_dir / f"{arxiv_id}.tar.gz"
        pdf_path = save_dir / f"{arxiv_id}.pdf"

        with open(tar_path, "wb") as f:
            f.write(eprint_resp.content)

        with open(pdf_path, "wb") as f:
            f.write(pdf_resp.content)

        return True, "ok"

    except requests.exceptions.Timeout:
        return False, "timeout"
    except Exception as e:
        return False, f"exception: {str(e)}"


def save_article_metadata(article_dir: Path, metadata: dict):
    """
    给每篇文章单独保存一个 metadata.json
    """
    meta_path = article_dir / "metadata.json"
    with open(meta_path, "w", encoding="utf-8") as f:
        json.dump(metadata, f, ensure_ascii=False, indent=2)


def save_global_csv(csv_path: Path, rows: list):
    """
    保存总 metadata.csv
    """
    if not rows:
        return

    fieldnames = list(rows[0].keys())
    with open(csv_path, "w", encoding="utf-8-sig", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(rows)


def save_summary_json(json_path: Path, summary: dict):
    with open(json_path, "w", encoding="utf-8") as f:
        json.dump(summary, f, ensure_ascii=False, indent=2)


def build_processing_order(normalized_mapping: dict):
    """
    按 大类 / 自定义子类 / 官方小类 排序，保证处理顺序稳定
    """
    records = []
    for official_cat, info in normalized_mapping.items():
        records.append((
            info["big_class"],
            info["custom_subclass"],
            official_cat
        ))
    return sorted(records, key=lambda x: (x[0], x[1], x[2]))


def build_dataset_article_id(global_index: int) -> str:
    """
    全局唯一编号
    例如：ART000001
    """
    return f"ART{global_index:06d}"


# =========================================================
# 3. 主流程
# =========================================================
def collect_arxiv_by_category(arxiv_category_mapping: dict):
    normalized_mapping = normalize_mapping(arxiv_category_mapping)
    group_schema = build_group_schema(normalized_mapping)
    official_targets = build_official_targets(normalized_mapping)
    processing_order = build_processing_order(normalized_mapping)

    client = arxiv.Client(
        page_size=CLIENT_PAGE_SIZE,
        delay_seconds=CLIENT_DELAY_SECONDS
    )

    session = requests.Session()
    session.headers.update(HEADERS)

    # 根目录
    base_dir = Path("data/archive/initial_collection") / FILE_NAME
    make_dir_if_not_exist(base_dir)

    # 放总 metadata
    meta_dir = base_dir / "_metadata"
    make_dir_if_not_exist(meta_dir)

    # 全局去重：同一篇文章只保留一次
    global_seen_ids = set()

    # 全局编号计数器
    global_article_counter = 0

    # 组内计数器：用于自定义子类内部编号
    group_article_counters = defaultdict(int)

    # 官方小类计数器：用于官方小类内部编号
    official_article_counters = defaultdict(int)

    # 总表
    all_metadata_rows = []

    # 汇总信息
    official_summary = {}
    group_summary = {
        f"{big_class} / {custom_subclass}": {
            "big_class": big_class,
            "custom_subclass": custom_subclass,
            "official_categories": official_categories,
            "final_count": 0
        }
        for (big_class, custom_subclass), official_categories in sorted(group_schema.items(), key=lambda x: (x[0][0], x[0][1]))
    }

    total_official_categories = len(processing_order)

    print(f"共检测到 {total_official_categories} 个官方小类。")
    print(f"下载根目录：{base_dir.resolve()}")
    print("-" * 100)

    for idx, (big_class, custom_subclass, official_cat) in enumerate(processing_order, start=1):
        target_count = official_targets[official_cat]
        official_count = 0

        group_key = (big_class, custom_subclass)
        group_dir = base_dir / sanitize_name(big_class) / sanitize_name(custom_subclass)
        make_dir_if_not_exist(group_dir)

        print(f"[{idx}/{total_official_categories}] 开始处理官方小类：{official_cat}")
        print(f"  归属目录：{big_class} / {custom_subclass}")
        print(f"  目标数量：{target_count}")

        search_stages = [
            ("date_range", True),
            ("fallback_all_time", False),
        ]

        for stage_name, use_date_range in search_stages:
            if official_count >= target_count:
                break

            print(f"  -> 当前阶段：{stage_name}")

            query = build_query(official_cat, use_date_range=use_date_range)

            search = arxiv.Search(
                query=query,
                max_results=MAX_RESULTS_PER_OFFICIAL_CATEGORY,
                sort_by=arxiv.SortCriterion.SubmittedDate,
                sort_order=arxiv.SortOrder.Descending
            )

            try:
                for result in client.results(search):
                    if official_count >= target_count:
                        break

                    arxiv_id = get_arxiv_id_from_result(result)

                    # 全局去重：已经被任何官方小类收录过，就直接跳过
                    if arxiv_id in global_seen_ids:
                        continue

                    # 当前官方小类必须至少在 categories 中出现
                    # 不再强制 primary_cat == official_cat，这样能更充分覆盖官方小类
                    if official_cat not in set(result.categories or []):
                        continue

                    ok, msg = download_tar_and_pdf(
                        session=session,
                        arxiv_id=arxiv_id,
                        save_dir=Path("__temp_placeholder__")
                    )

                    if not ok:
                        continue

                    # 真正通过后再分配编号、建最终目录、写文件
                    global_article_counter += 1
                    official_article_counters[official_cat] += 1
                    group_article_counters[group_key] += 1

                    dataset_article_id = build_dataset_article_id(global_article_counter)
                    official_index = official_article_counters[official_cat]
                    group_index = group_article_counters[group_key]

                    article_dir = group_dir / f"{dataset_article_id}_{arxiv_id}"

                    # 把临时测试下载替换为正式下载
                    # 为避免重复网络请求，这里直接重新下载到最终目录
                    # 如果正式下载失败，则回滚编号计数
                    ok_final, msg_final = download_tar_and_pdf(session, arxiv_id, article_dir)
                    if not ok_final:
                        global_article_counter -= 1
                        official_article_counters[official_cat] -= 1
                        group_article_counters[group_key] -= 1
                        continue

                    primary_cat = result.primary_category
                    title = result.title.strip().replace("\n", " ")
                    authors = ", ".join([str(a) for a in result.authors]) if result.authors else ""
                    abstract = result.summary.strip().replace("\n", " ") if result.summary else ""

                    row = {
                        "dataset_article_id": dataset_article_id,
                        "global_index": global_article_counter,
                        "group_index": group_index,
                        "official_index": official_index,
                        "arxiv_id": arxiv_id,
                        "title": title,
                        "authors": authors,
                        "published": str(result.published) if result.published else "",
                        "updated": str(result.updated) if result.updated else "",
                        "official_query_category": official_cat,
                        "official_primary_category": primary_cat,
                        "all_categories": ";".join(result.categories) if result.categories else "",
                        "big_class": big_class,
                        "custom_subclass": custom_subclass,
                        "entry_id": result.entry_id,
                        "pdf_url": f"https://arxiv.org/pdf/{arxiv_id}.pdf",
                        "abstract": abstract
                    }

                    save_article_metadata(article_dir, row)

                    all_metadata_rows.append(row)
                    global_seen_ids.add(arxiv_id)
                    official_count += 1
                    group_summary[f"{big_class} / {custom_subclass}"]["final_count"] += 1

                    print(
                        f"    √ {dataset_article_id} | {arxiv_id} | "
                        f"official {official_count}/{target_count} | "
                        f"group #{group_index}"
                    )

            except Exception as e:
                print(f"    ! 查询 {official_cat} 时出错：{str(e)}")
                continue

        official_summary[official_cat] = {
            "official_category": official_cat,
            "big_class": big_class,
            "custom_subclass": custom_subclass,
            "target_count": target_count,
            "final_count": official_count,
            "status": "OK" if official_count >= target_count else "LOW"
        }

        print(
            f"  完成：{official_cat} -> {big_class} / {custom_subclass} "
            f"共 {official_count}/{target_count} 篇"
        )
        print("-" * 100)

    # 保存总 metadata.csv
    csv_path = meta_dir / "all_articles_metadata.csv"
    save_global_csv(csv_path, all_metadata_rows)

    # 保存汇总 summary.json
    summary_path = meta_dir / "summary.json"
    summary_payload = {
        "config": {
            "file_name": FILE_NAME,
            "date_start": DATE_START,
            "date_end": DATE_END,
            "official_target_choices": list(OFFICIAL_TARGET_CHOICES),
            "random_seed": RANDOM_SEED,
            "max_results_per_official_category": MAX_RESULTS_PER_OFFICIAL_CATEGORY,
            "article_time_limit_seconds": ARTICLE_TIME_LIMIT_SECONDS
        },
        "totals": {
            "official_category_count": len(normalized_mapping),
            "custom_group_count": len(group_schema),
            "total_effective_articles": len(all_metadata_rows)
        },
        "official_summary": official_summary,
        "group_summary": group_summary
    }
    save_summary_json(summary_path, summary_payload)

    # 再保存一个简单统计文本
    txt_path = meta_dir / "summary.txt"
    with open(txt_path, "w", encoding="utf-8") as f:
        total_articles = len(all_metadata_rows)
        f.write(f"总有效文章数：{total_articles}\n")
        f.write(f"官方小类数：{len(normalized_mapping)}\n")
        f.write(f"自定义子类数：{len(group_schema)}\n\n")

        f.write("=== 官方小类统计 ===\n")
        for official_cat in sorted(official_summary.keys()):
            v = official_summary[official_cat]
            f.write(
                f"{official_cat}: {v['final_count']} / target={v['target_count']} "
                f"-> {v['big_class']} / {v['custom_subclass']} / status={v['status']}\n"
            )

        f.write("\n=== 自定义子类统计 ===\n")
        for k in sorted(group_summary.keys()):
            v = group_summary[k]
            f.write(f"{k}: {v['final_count']}\n")

    print("\n全部任务完成。")
    print(f"总有效文章数：{len(all_metadata_rows)}")
    print(f"总 metadata：{csv_path}")
    print(f"汇总文件：{summary_path}")


# =========================================================
# 4. 运行入口
# =========================================================
if __name__ == "__main__":
    collect_arxiv_by_category(ARXIV_CATEGORY_MAPPING)
```
